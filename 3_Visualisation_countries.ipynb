```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url="https://raw.githubusercontent.com/plotly/datasets/master/gapminderDataFiveYear.csv"

df=pd.read_csv(url)


print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
```

```python
df = df.rename(columns={
    "lifeExp": "life_expectancy",
    "pop": "population",
    "gdpPercap": "gdp_per_capita"
})

print(df.head())
print(df.columns)
```

```python
print("Missing values in each column:")
print(df.isna().sum)

print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df_clean = df.copy()


if df_clean.duplicated().sum() > 0:
    df_clean = df_clean.drop_duplicates()


for col in ["country", "continent"]:
    df_clean[col] = df_clean[col].astype(str).str.strip()


df_clean["year"] = df_clean["year"].astype(int)
df_clean["population"] = df_clean["population"].astype(float)
df_clean["life_expectancy"] = df_clean["life_expectancy"].astype(float)
df_clean["gdp_per_capita"] = df_clean["gdp_per_capita"].astype(float)

print(df_clean.dtypes)
print(df_clean.head())

```

```python
dirty_demo = df_clean.head(12).copy()


dirty_demo.loc[2, "life_expectancy"] = np.nan
dirty_demo.loc[5, "gdp_per_capita"] = np.nan

print(dirty_demo)
print(dirty_demo.isna().sum())


clean_demo = dirty_demo.copy()
clean_demo["life_expectancy"] = clean_demo["life_expectancy"].fillna(clean_demo["life_expectancy"].median())
clean_demo["gdp_per_capita"] = clean_demo["gdp_per_capita"].fillna(clean_demo["gdp_per_capita"].median())

print(clean_demo.isna().sum())
print(clean_demo)
```

```python
min_year = df['year'].min()
max_year = df['year'].max()
print(f"The year range in the dataset is from {min_year} to {max_year}.")


num_countries = df['country'].nunique()
print("Number of Countries:", num_countries)


num_continents = df['continent'].nunique()
print("Number of Continents:", num_continents)
```

```python
latest_year = df_clean['year'].max()

latest_year_df = df_clean[df_clean['year'] == latest_year]

continent_life = latest_year_df.groupby('continent')['life_expectancy'].mean().sort_values(ascending=True)

plt.figure(figsize=(10, 6))
plt.bar(continent_life.index, continent_life.values, color='skyblue')

plt.title(f'Average Life Expectancy by Continent in {latest_year}', fontsize=14)
plt.xlabel('Continent', fontsize=12)
plt.ylabel('Average Life Expectancy', fontsize=12)

plt.xticks(rotation=45, ha='right')
plt.yticks(fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python


countries = ["India", "China", "United States"]

filtered_df = df[df["country"].isin(countries)]

for country in countries:
    country_data = filtered_df[filtered_df["country"] == country]
    plt.plot(country_data["year"],
             country_data["life_expectancy"],
             label=country)

plt.title("Life Expectancy Over Time")
plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df_clean['gdp_per_capita'], df_clean['life_expectancy'], alpha=0.7)

plt.title('GDP Per Capita vs. Life Expectancy', fontsize=14)
plt.xlabel('GDP Per Capita', fontsize=12)
plt.ylabel('Life Expectancy', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
