```python
# Calculate average marks and print class passed/failed
marks = [45, 60, 72, 38, 90]

total = 0

for i in range(len(marks)):
    total += marks[i]

average = total / len(marks)

if average > 50:
    print("Class passed")
else:
    print("Class failed")

print("Average:", average)
```

```python
import pandas as pd

data = {
    "student_id": [101, 102, 103, 104, 105, 106],
    "attendance_percent": [92, 67, 81, 45, 74, 88],
    "assignment_score": [18, 12, 15, 8, 14, 19],
    "quiz_score": [72, 48, 65, 30, 55, 80],
    "lab_completed": [True, False, True, False, True, True]
}

df = pd.DataFrame(data)
df

```

```python
df['total_score'] = df['assignment_score'] + df['quiz_score']

df
```

```python
df['eligible'] = (
    (df['attendance_percent'] >= 75) &
    (df['total_score'] >= 70) &
    (df['lab_completed'])
)

df
```

```python
eligible_students = df[df["eligible"] == True]

eligible_students
```

```python
# Performance percentage
df["performance_percent"] = (df["total_score"] / 100) * 100

# Attendance category
df["attendance_category"] = df["attendance_percent"].apply(
    lambda x: "Good" if x >= 75 else "Poor"
)

# Grade
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "D"

df["grade"] = df["total_score"].apply(grade)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(df["student_id"], df["total_score"])
plt.title("Total Score of Students")
plt.xlabel("Student ID")
plt.ylabel("Total Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["student_id"], df["attendance_percent"], marker="o")
plt.title("Attendance Percentage")
plt.xlabel("Student ID")
plt.ylabel("Attendance %")
plt.show()
```

```python
df["eligible"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Eligible Students Distribution")
plt.ylabel("")
plt.show()
```

```python
df["grade"].value_counts().plot(kind="bar")
plt.title("Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Count")
plt.show()
```
