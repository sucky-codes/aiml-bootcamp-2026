```python
from google.colab import files
import pandas as pd
import numpy as np

uploaded = files.upload()

df = pd.read_csv("session4_dummy_students_data.csv.xls")

print(df)
```

```python
math_marks = np.array(df["math"])
science_marks = np.array(df["science"])
english_marks = np.array(df["english"])

average = (math_marks + science_marks + english_marks) / 3

df["average"] = average

print(df[["name","average"]])
```

```python
result = []

for mark in average:
    if mark >= 40:
        result.append("Pass")
    else:
        result.append("Fail")

df["result"] = result

print(df[["name","average","result"]])
```

```python
class_average = df["average"].mean()

print("Class Average =", class_average)
```

```python
highest_average = df["average"].max()
lowest_average = df["average"].min()

print("Highest Average =", highest_average)
print("Lowest Average =", lowest_average)
```

```python
highest_average = df["average"].max()

top_students = df[df["average"] == highest_average]

print("Top Student(s):")
print(top_students)
```

```python
low_attendance = df[df["attendance_percent"] < 75]

print("Students with attendance below 75:")
print(low_attendance)
```

```python
few_assignments = df[df["assignments_submitted"] < 6]

print("Students who submitted fewer than 6 assignments:")
print(few_assignments)
```

```python
df.to_csv("session4_student_report.csv", index=False)

print("Final Report:")
print(df)

print("session4_student_report.csv has been saved.")
```
